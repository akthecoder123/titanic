```python
df['Age'].fillna(df['Age'].median(),inplace=True)
df['Cabin'].fillna("NotSpecified",inplace=True)
df['Embarked'].fillna("Na",inplace=True)
```

```python
sns.countplot(x='Survived',data=df)
```

```python
sns.clustermap(pd.crosstab(df['Survived'],df['Pclass']))
```

```python
survival_percentage=df.groupby('Pclass')['Survived'].mean()*100
survival_percentage
```

```python
survival_percentage.plot(kind='bar')
```

```python
df=pd.get_dummies(df,drop_first=True)
df
```

```python
x=df.drop(['Survived'],axis=1)
```

```python
y=df['Survived']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.1)
```

```python
from sklearn.preprocessing import StandardScaler
scale=StandardScaler()
x_train=scale.fit_transform(x_train)
x_test=scale.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
clf=LogisticRegression()
clf.fit(x_train,y_train)
```

```python
y_predict=clf.predict(x_test)
```

```python
y_test
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_predict)
```
